## Meet the List

So, what exactly is a list? A list is a collection of items stored in a single variable: one name, holding many values. Instead of fruit one, fruit two, fruit three, you keep one list called fruits, and the items live inside it, side by side. Feature one: a list is ordered. The items have a fixed position: first, second, third, and that order does not change unless you change it. Feature two: a list is changeable. After creating it, you can add items, remove items, and replace items whenever you like. Feature three: duplicates are allowed. The same value may appear twice, three times, as often as your data needs. And feature four: a list can hold any data type. Strings, numbers, booleans — even a mix of all of them inside one single list. That combination, ordered and changeable, makes the list the default container for almost everything in Python.

- **ORDERED** — Fixed positions: 1st, 2nd, 3rd
- **CHANGEABLE** — Add, remove, replace freely
- **DUPLICATES ALLOWED** — Same value, many times
- **ANY TYPE INSIDE** — Text, numbers, booleans, mixed

**One variable — many values**

## Create a List

Creating a list takes one pair of square brackets. Inside the brackets, write your items, separated by commas. Here is a list of three fruits: apple, banana, cherry, stored in a variable called thislist. The brackets are the signal: square brackets mean list, just as quotes mean string. Notice the items keep the order you wrote them: apple first, banana second, cherry third. Print the list, and Python shows it back with its brackets and quotes: apple, banana, cherry. Ask for its length with len, and the answer is three: len counts the items inside. Three items, counted — and len will keep counting correctly as the list grows and shrinks today. Ask for its type, and Python answers: class list — a brand new data type in your toolbox. There is also a second way to build the same list: the list constructor. Write list, then double round brackets around a tuple of items: the inner pair holds the values, the outer pair makes the call. The result is exactly the same list. Both roads lead to the same place; the bracket way is the one you will type ninety nine times out of a hundred. And remember feature four: a list can mix types — here is text, a number, a boolean, another number and text again, all in one list. Printed exactly as stored: abc, thirty four, True, forty, male. And duplicates? Watch apple and cherry appear twice in this list. No complaint at all: the list keeps both copies, in order.

- *Note: Square brackets [ ] = a list*

```python
thislist = ["apple", "banana", "cherry"]
print(thislist)
print(len(thislist))
print(type(thislist))
```

## Access Items by Index

A list keeps its items in order — so how do you reach one item? By its index: the position number of the item inside the list. Here is our fruit list again. Now the rule that surprises every beginner: Python counts positions starting from zero, not from one. Index zero is the first item: apple. Index one is the second item: banana. And index two is the third: cherry. Square brackets with a number inside: that is how you ask a list for one position. There is also a second counting system, from the end: negative indexing. Negative indexes count backwards, starting at minus one. Minus one is always the last item: cherry — the quickest way to grab the end of any list, whatever its length. And minus two is the second from the end: banana. Positive from the front, negative from the back: two doors into the same list.

- *Note: Index starts at 0 — the #1 beginner surprise*

```python
thislist = ["apple", "banana", "cherry"]
print(thislist[0])
print(thislist[1])
print(thislist[2])
```

```python
thislist = ["apple", "banana", "cherry"]
print(thislist[-1])
print(thislist[-2])
```

## Ranges of Indexes

One item at a time is useful, but often you want a slice of the list. Python answers with a range of indexes: start, colon, end. Our list grows to seven fruits: apple, banana, cherry, orange, kiwi, melon, mango — indexes zero to six. A range builds a brand new, smaller list from the positions you choose. Start at index two, end at index five. The result: cherry, orange, kiwi. Memorise this: the end index is NOT included — the slice stops one position before it, so index five, melon, stays out. Leave the start empty, and the slice begins at the very beginning: the first four items. Apple, banana, cherry, orange. Leave the end empty instead, and the slice runs from your start to the end of the list: cherry, orange, kiwi, melon, mango. Ranges work with negative indexes too. Minus four to minus one. Counting from the back: orange, kiwi, melon — and again, the end position itself is left out. One last access trick: checking whether an item exists at all, with the word in. If apple is in thislist, print yes — and it is, so the message appears. Ranges give you sub-lists; in gives you a yes or no. Both will be daily tools from now on.

- *Note: Range end is EXCLUDED: [2:5] stops at index 4*

```python
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(thislist[2:5])
print(thislist[:4])
print(thislist[2:])
```

```python
print(thislist[-4:-1])
if "apple" in thislist:
  print("Yes, 'apple' is in the fruits list")
```

## Change Item Values

Changeable means changeable: items inside a list can be replaced. Aim at one item by its index, and assign a new value. Our list: apple, banana, cherry. thislist, index one, equals blackcurrant. Banana is gone, blackcurrant sits in its place, and the list keeps its size and order. You can also replace a whole range in one move. Six fruits this time. Replace indexes one to three — that is banana and cherry — with blackcurrant and watermelon. Two items out, two items in: the list flows around the change like water. Now the two surprises hiding in this trick. Fresh list: apple, banana, cherry. Replace just one item, index one to two, with two new items. The extra item inserts itself, and the list grows: apple, blackcurrant, watermelon, cherry. And the mirror image works too. Fresh list again. Replace two items, indexes one to three, with only one item: watermelon. The list shrinks to two items: apple and watermelon. So a range assignment is also a hidden insert and a hidden delete. Powerful — so read such lines slowly.

- *Note: Range assignment can insert AND delete*

```python
thislist = ["apple", "banana", "cherry"]
thislist[1] = "blackcurrant"
print(thislist)
```

```python
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "mango"]
thislist[1:3] = ["blackcurrant", "watermelon"]
print(thislist)
```

```python
thislist = ["apple", "banana", "cherry"]
thislist[1:2] = ["blackcurrant", "watermelon"]
print(thislist)
```

```python
thislist = ["apple", "banana", "cherry"]
thislist[1:3] = ["watermelon"]
print(thislist)
```

## Add Items: append · insert · extend

Now let us grow a list: adding new items. We start with our three fruits: apple, banana, cherry. Tool number one: append. append takes one item and adds it to the very end of the list. No position to choose, no confusion: the end, always. fruits dot append, orange — and orange slides in at the end, at index three. Tool number two: insert, for adding at an exact position. insert takes two things: first the index where the item belongs, then the item itself. Insert lemon at index one — watch banana, cherry and orange shuffle one step to the right to make room. Tool number three: extend, for adding many items at once. Here is a second list called tropical: mango, pineapple, papaya. fruits dot extend, tropical — and all three items pour into fruits, keeping their order. And extend is generous: it accepts any iterable, not only lists. Extend with a tuple — kiwi and melon — and they join the list just the same. Ten items from three, with three small tools: append for the end, insert for a position, extend for many at once.

- *Note: append = end · insert = position · extend = many*

```python
fruits = ["apple", "banana", "cherry"]
print(fruits)
```

```python
fruits.append("orange")
print(fruits)
```

```python
fruits.insert(1, "lemon")
print(fruits)
```

```python
tropical = ["mango", "pineapple", "papaya"]
fruits.extend(tropical)
print(fruits)
```

```python
fruits.extend(("kiwi", "melon"))
print(fruits)
```

## Remove Items: remove · pop · del · clear

Lists shrink as easily as they grow: four tools remove items. Our list: apple, banana, cherry, banana, orange — and yes, banana appears twice, on purpose. Tool one: remove, which deletes an item by its value. fruits dot remove, banana. Watch closely: only the first banana vanished — the one at index one. remove always deletes the first matching item it finds. And if the value is not in the list at all, Python refuses with a ValueError: list dot remove, x: x not in list. Tool two: pop. pop removes an item by index, and hands the removed item back to you. pop, index one: cherry comes out, and we catch it in the variable x. Returning the item makes pop perfect for queues and stacks, where you process items one by one. Call pop with no index at all, and it takes the last item instead: orange pops off, into y. Tool three: the del keyword. del deletes by index too, but returns nothing: the item is simply gone. del fruits, index zero: apple disappears, silently. del can even delete the entire list, name and all — print it afterwards, and Python answers with a NameError. And tool four: clear. fruits dot clear empties the list completely — every item out. But notice: the list itself survives, empty and ready to be reused. That is the difference between clear and del. remove by value, pop by index, del by index, clear for everything: a full shrinking toolkit.

- *Note: remove = by value · pop/del = by index · clear = all*

```python
fruits = ["apple", "banana", "cherry", "banana", "orange"]
print(fruits)
```

```python
fruits.remove("banana")
print(fruits)
```

```python
x = fruits.pop(1)
print(x)
print(fruits)
```

```python
y = fruits.pop()
print(y)
print(fruits)
```

```python
del fruits[0]
print(fruits)
```

```python
fruits.clear()
print(fruits)
```

## Loop Through a List

Lists and loops are best friends: you will constantly walk through a list, item by item. Same fruit list. The cleanest way is the for loop: for x in thislist, print x. Python takes each item in turn — apple, banana, cherry — puts it in x, and the body runs once per item. No counters, no indexes to manage: the loop ends by itself when the items run out. Sometimes you need the positions as well; then loop over the indexes. for i in range of len of thislist: i becomes zero, one, two. And print thislist at index i gives the same three fruits, this time reached by their numbers. Same output — but now you hold the index, ready to change items or compare neighbours. Third way: the while loop, the manual gearbox. Start a counter i at zero, and loop while i is smaller than the length of the list. Print the item at i — and never forget the last line: i equals i plus one. Forget that line, and i never grows: the loop prints apple forever. That is an infinite loop, and every programmer has built one by accident. Three loops, one list: for each item directly, for each index, or while with a counter you control.

- *Note: Forget i = i + 1 → infinite loop!*

```python
thislist = ["apple", "banana", "cherry"]
for x in thislist:
  print(x)
```

```python
thislist = ["apple", "banana", "cherry"]
for i in range(len(thislist)):
  print(thislist[i])
```

```python
thislist = ["apple", "banana", "cherry"]
i = 0
while i < len(thislist):
  print(thislist[i])
  i = i + 1
```

## List Comprehension

Now Python's famous party trick: list comprehension — a brand new list, built in a single line. Our fruits: apple, banana, cherry, kiwi, mango. The goal: a new list holding only the fruits whose name contains the letter a. The long way: start empty, loop, test each fruit with an if, and append the keepers. Result: apple, banana, mango. Cherry and kiwi contain no letter a, so they are left behind. Six lines for one small filter. Now the same job in a single line. Same fruits, same goal. newlist equals: bracket — x — for x in fruits — if a in x — bracket. Read it as a sentence: give me x, for every x in fruits, if a is in x. Identical result: apple, banana, mango. That is the grammar of every comprehension: an expression, a for loop, and an optional if — inside square brackets. Drop the if, and the comprehension copies the whole list, item by item. Or change the condition: x not equal to apple filters out exactly one fruit. Banana, cherry, kiwi and mango survive. What you loop over — the iterable — can be any loopable thing, not only a list. Loop over range ten, keep the numbers below five: zero, one, two, three, four. And the expression at the front can transform each item as it passes. x dot upper shouts every fruit in capitals: APPLE, BANANA, CHERRY, KIWI, MANGO. The expression can even carry a condition of its own. x — if x is not banana — else orange: everything passes unchanged, except banana, which is swapped for orange. Apple, orange, cherry, kiwi, mango. Short, readable, fast: when a loop's only job is building a list, the comprehension is usually the better tool.

- *Note: [ expression for item in iterable if condition ]*

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = []
for x in fruits:
  if "a" in x:
    newlist.append(x)
print(newlist)
```

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x for x in fruits if "a" in x]
print(newlist)
newlist = [x for x in fruits]
print(newlist)
newlist = [x for x in fruits if x != "apple"]
print(newlist)
```

```python
newlist = [x for x in range(10) if x < 5]
print(newlist)
newlist = [x.upper() for x in fruits]
print(newlist)
newlist = [x if x != "banana" else "orange" for x in fruits]
print(newlist)
```

## Sort Lists

A list rarely arrives in the order you want — so Python sorts it in place. Five fruits, deliberately scrambled: orange, mango, kiwi, pineapple, banana. thislist dot sort, and the list rearranges itself alphabetically. Banana, kiwi, mango, orange, pineapple — ascending order, smallest first. Numbers sort numerically, of course: one hundred, fifty, sixty five, eighty two, twenty three becomes twenty three up to one hundred. And sorting works directly on the list itself — there is no new list; the original changes. Remember that. Want the opposite direction? Same scrambled list. sort with reverse equals True flips the order: pineapple down to banana. And the numbers run from one hundred down to twenty three. Now a famous trap, hiding in mixed capitals. banana with a small b, Orange and Kiwi with capitals, cherry with a small c. Sort it — and the result looks wrong: Kiwi, Orange, banana, cherry. It is not wrong: sort is case sensitive, and in the computer's alphabet every capital letter comes before every small letter. The fix is a key: sort with key equals str dot lower tells Python, compare the items as if they were all lowercase. Now the order is truly alphabetical: banana, cherry, Kiwi, Orange. A key is a function Python runs on every item before comparing — and you can write your own. This key returns the distance from fifty: absolute value of n minus fifty. Sort the numbers by that key, and they line up by closeness to fifty. Fifty itself first, then sixty five, twenty three, eighty two, and finally one hundred. Last tool: reverse simply flips a list backwards, whatever order it is in. Cherry, Kiwi, Orange, banana — the exact mirror image.

- *Note: sort() changes the list itself — no new list*

```python
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort()
print(thislist)
numbers = [100, 50, 65, 82, 23]
numbers.sort()
print(numbers)
```

```python
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort(reverse = True)
print(thislist)
numbers = [100, 50, 65, 82, 23]
numbers.sort(reverse = True)
print(numbers)
```

```python
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort()
print(thislist)
thislist.sort(key = str.lower)
print(thislist)
```

```python
def myfunc(n):
  return abs(n - 50)

thislist = [100, 50, 65, 82, 23]
thislist.sort(key = myfunc)
print(thislist)
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.reverse()
print(thislist)
```

## Copy and Join Lists

You will often need a copy of a list — and here sits the most famous beginner trap in Python. Our three fruits. mylist equals thislist: it looks like a copy. Append orange to mylist. Now print thislist — and orange is in it too! We never touched thislist, yet it changed. The equals sign did not copy the list: it copied the address. Both names point at the same single list. The real tool is the copy method: mylist equals thislist dot copy — a fresh, independent list with the same items. Append kiwi to the copy. This time thislist keeps only its four items, while mylist has five: the copy is truly separate. Rule of thumb: equals shares, copy duplicates. When in doubt, copy. Two more copying tools, quickly: the list constructor makes a copy. list of thislist: same items, new list. And the slice trick: thislist, colon inside brackets, copies the whole list — a shortcut you will see in older code. Now the opposite job: joining lists into one. list1 holds the letters a, b, c; list2 holds the numbers one, two, three. Plus joins them: list3 is list1 plus list2 — letters then numbers, one new list. Joining never changes the two originals; it builds a third list. Second way: loop and append. For each x in list2, append it to list1 — same joined result, built by hand. And the third way is the one you already know: extend does the same job in one call.

- *Note: = shares the list · .copy() duplicates it*

```python
thislist = ["apple", "banana", "cherry"]
mylist = thislist
mylist.append("orange")
print(thislist)
mylist = thislist.copy()
mylist.append("kiwi")
print(thislist)
print(mylist)
```

```python
thislist = ["apple", "banana", "cherry"]
mylist = list(thislist)
print(mylist)
mylist = thislist[:]
print(mylist)
```

```python
list1 = ["a", "b", "c"]
list2 = [1, 2, 3]
list3 = list1 + list2
print(list3)
```

```python
list1 = ["a", "b", "c"]
list2 = [1, 2, 3]
for x in list2:
  list1.append(x)
print(list1)
list1 = ["a", "b", "c"]
list1.extend(list2)
print(list1)
```

## Meet the Tuple

Collection number two: the tuple. A tuple is a list's closest relative: a collection of items, in one variable, written with round brackets instead of square ones. It shares most of the list's nature. It is ordered: items keep their positions, and you reach them by index, exactly as before. But then comes the one giant difference. A tuple is unchangeable. Once a tuple is created, you cannot add items, remove items, or change items. What was born in the tuple, stays in the tuple. It also allows duplicates, like its twin. And it happily holds any data type, mixed or single. Strings, numbers, booleans — one tuple can carry them all. Why would anyone choose a container that refuses to change? Because some data must NOT change: the days of the week, coordinates on a map, the red green blue of a colour. A tuple is Python's way of promising: this data is fixed. And Python enforces the promise.

- **ORDERED** — Same indexes as lists
- **UNCHANGEABLE** — No add, no remove, no edit
- **DUPLICATES ALLOWED** — Same value, many times
- **ANY TYPE INSIDE** — Text, numbers, booleans, mixed

**Like a list — but locked**

## Create Tuples

Creating a tuple looks almost identical to creating a list. Round brackets, items separated by commas: thistuple holds apple, banana, cherry. Print it, and the round brackets in the output confirm: this is a tuple. len answers three, exactly as with lists. Now the smallest trap in all of Python — the one item tuple. A tuple of one item needs its comma: apple, comma, inside the brackets. With the comma, the type is tuple. Forget the comma, and the brackets are just quiet decoration around a string. The type answers str — not a tuple at all. One comma makes the difference. Types inside a tuple are as free as in a list. Strings in tuple1, numbers in tuple2, booleans in tuple3 — all at home. Printed one by one, exactly as stored. And tuples have a constructor too: tuple, double brackets — the same pattern as list.

- *Note: One item? The COMMA makes the tuple: ("apple",)*

```python
thistuple = ("apple", "banana", "cherry")
print(thistuple)
print(len(thistuple))
```

```python
thistuple = ("apple",)
print(type(thistuple))
notatuple = ("apple")
print(type(notatuple))
```

```python
tuple1 = ("apple", "banana", "cherry")
tuple2 = (1, 5, 7, 9, 3)
tuple3 = (True, False, False)
print(tuple1)
print(tuple2)
print(tuple3)
thistuple = tuple(("apple", "banana", "cherry"))
print(thistuple)
```

## Update Tuples (the Workarounds)

Now the moment of truth: can you change a tuple? thistuple, apple banana cherry — try to replace index one with blackcurrant, exactly as we did with lists. The assignment line looks perfectly reasonable. Python slams the door: TypeError — tuple object does not support item assignment. No appeal, no warning: tuples simply refuse. Unchangeable means unchangeable. But real life still needs fixes — so Python programmers use a famous workaround. Step one: convert the tuple into a list with the list function. y is now a list twin of x. Step two: change the list freely — index one becomes kiwi. Step three: convert back with tuple. And x is now apple, kiwi, cherry: the tuple changed — by being rebuilt. The same spirit adds items. thistuple again, and a second tuple y holding one orange — comma included, as you learned. Tuples cannot grow, but two tuples can join: thistuple plus-equals y. The plus-equals builds a brand new tuple and parks it under the same name. Result: apple, banana, cherry, orange — four items. Removing follows the list-conversion road as well. Convert to a list. remove apple from the list, then convert back. thistuple is now just banana and cherry. And the one destructive power a tuple does allow: its own deletion. del thistuple deletes the whole tuple; print it afterwards... ...and Python answers NameError: name thistuple is not defined. It is gone entirely.

- *Note: Convert → change → convert back*

```python
thistuple = ("apple", "banana", "cherry")
thistuple[1] = "blackcurrant"
```

```python
x = ("apple", "banana", "cherry")
y = list(x)
y[1] = "kiwi"
x = tuple(y)
print(x)
```

```python
thistuple = ("apple", "banana", "cherry")
y = ("orange",)
thistuple += y
print(thistuple)
```

```python
thistuple = ("apple", "banana", "cherry")
y = list(thistuple)
y.remove("apple")
thistuple = tuple(y)
print(thistuple)
del thistuple
print(thistuple)
```

## Unpack Tuples

In Part One you unpacked a list into variables. Tuples unpack the same way — in fact, unpacking is a tuple's home ground. fruits is a tuple: apple, banana, cherry. Write variables in brackets on the left — green, yellow, red — and assign the tuple. Python deals the values out in order. green is apple, yellow is banana, red is cherry — printed one by one to prove it. Five fruits now, and only three variables: the counts do not match. fruits: apple, banana, cherry, strawberry, raspberry. Mark the last variable with a star: star red. The star changes everything: red does not take one value... green takes apple, yellow takes banana... ...and red collects ALL the rest — cherry, strawberry, raspberry — as a list. Notice the square brackets in its output. The star can also stand in the middle. apple, mango, papaya, pineapple, cherry. green, star tropic, red. green takes the first, red takes the last... ...and tropic gathers everything in between: mango, papaya, pineapple — again as a list. One star per unpacking, placed wherever the flexible part of your data sits.

- *Note: * starred variable always becomes a LIST*

```python
fruits = ("apple", "banana", "cherry")
(green, yellow, red) = fruits
print(green)
print(yellow)
print(red)
```

```python
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")
(green, yellow, *red) = fruits
print(green)
print(yellow)
print(red)
```

```python
fruits = ("apple", "mango", "papaya", "pineapple", "cherry")
(green, *tropic, red) = fruits
print(green)
print(tropic)
print(red)
```

## Loop, Join and Tuple Methods

Last stretch: loops, joining, and the tuple's only two methods. thistuple, three fruits. Looping is identical to lists: for x in thistuple, print x. Apple, banana, cherry, one per line. And the index loop works too: for i in range of len, print thistuple at i. Same three fruits again — reached by their positions this time. Joining uses the plus sign, just like lists. tuple1 holds a, b, c; tuple2 holds one, two, three. tuple3 equals tuple1 plus tuple2: a, b, c, one, two, three. And tuples have one party trick: multiplication. fruits, the tuple. mytuple equals fruits times two: the whole tuple repeated — apple, banana, cherry, apple, banana, cherry. That leaves the methods — and a tuple has only two. thistuple is ten numbers, with five appearing twice. count answers how many times a value appears: count of five... ...is two. And index answers where a value first sits: index of eight is three. Two methods only — count and index — because a tuple never changes, it needs no tools for changing.
