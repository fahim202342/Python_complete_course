## Syntax: Line by Line

Now let us learn how Python code actually works: its syntax. Syntax means the rules for how code must be written, and Python's rules are famously gentle. Python executes your file line by line, from top to bottom. And unlike many languages, a Python statement needs no special ending: no semicolon, no marker, just the line itself. Look at these three lines. Each one is a complete statement, and Python runs them in order, first, second, third, exactly as you read them. The output appears in the same order: Hello World! ## Syntax: Indentation Is Law

Now the single most important syntax rule in Python: indentation. In most languages, the spaces at the start of a line are just decoration. In Python, those spaces are the law: indentation tells Python which lines belong together inside a block. Where other languages use curly brackets to group lines, Python uses the spaces themselves, which is why Python code looks so clean. Look at this example. If five is greater than two, colon: that line asks a question. The next line is pushed in, indented with spaces. That indented line is the block, and it runs only when the condition is true. Five is greater than two, so the message prints: Five is greater than two! Now watch what happens when we delete the indentation. Same words, same line, but no spaces before it. Python refuses to run the program at all. It complains with an IndentationError: expected an indented block after the if statement on line one. Beginners fear this error, but do not: it is Python protecting the structure of your code. It would rather stop and tell you, than guess what you meant and run the wrong thing. The fix is simple: indent the block properly, usually with four spaces. And keep every line in the same block at exactly the same depth, or Python will complain again. Remember: in Python, indentation is not style, it is syntax.

- *Note: Rule of thumb: 4 spaces for one block level*

````python
if 5 > 2:
  print("Five is greater than two!")
````

## Comments: Notes for Humans

Programs grow big fast, and future you will not remember why present you wrote a line. That is what comments are for. A comment starts with the hash symbol. Everything after the hash is ignored by Python: comments exist only for the humans reading the code. A comment can stand on its own line, above the code it explains. Or it can sit at the end of a code line, explaining just that line. Either way the program runs exactly the same, and Hello, World still prints. Comments are also a safe way to switch a line off while testing: put a hash in front of it, and Python skips it, until you remove the hash again. For several lines, you can put a hash at the start of each line: three lines, three hashes. Or use the popular trick: a multi-line string, three quotes above and three quotes below your text. Python reads it as a string that is never used, so nothing happens. Strictly speaking, that is a string and not a true comment, but you will see it in real code everywhere. Comments never change the program; they only save the reader. Write them like someone else must understand your code at midnight, because that someone is you.

- *Note: Comments = notes for humans — Python skips them*

````python
#This is a comment
print("Hello, World!")
````

````python
print("Hello, World!") #This is a comment
````

````python
#This is a comment
#written in
#more than just one line
print("Hello, World!")
````

````python
"""
This is a comment
written in
more than just one line
"""
print("Hello, World!")
````

## Variables: Named Boxes

Now the third foundation, one you will use in every program you ever write: variables. A variable is a container for storing a data value: picture a box with a name on it, holding something inside. Whenever your program needs to remember something, a score, a name, a total, it goes into a variable. Creating one in Python could not be simpler. Write the name, an equals sign, and the value: x equals five. Done: the variable x now holds the number five. Now y equals John, written in quotes. The variable y now holds the text John. The quotes matter: they are what tell Python, this is text, not a number and not another variable. Notice what did not happen: there is no declaration step and no special command. A Python variable is created the moment you first assign it a value. Reading the value back is just as easy. Print x, and you get five. Print y, and you get John. Python also remembers what kind of data each box holds. Ask with the type function, and it answers: class int for x, a whole number, and class str for y, a piece of text. Hold that picture: named boxes, each holding one value. Everything in Python builds on it.

- *Note: No declaration — assignment creates the variable*

````python
x = 5
y = "John"
print(x)
print(y)
print(type(x))
print(type(y))
````

## Variables Can Change Type

Here is something that surprises people coming from other languages. A Python variable is not married to one type. Watch this box called x. First, x equals four: x holds a number. Then, on the very next line, x equals Sally, a string. No complaint, no error. Print x now, and out comes Sally. The box kept its name and simply took a new value of a completely different type. This freedom is called dynamic typing, and it makes Python fast to write. It also means you must keep track of what your boxes currently hold. If you expect a number and the box quietly holds text, your program will trip over it later, and Python will not warn you in advance. And when you do care about the type, you can force it: that is called casting. str of three makes the text three. int of three makes the number three. float of three makes three point zero. We will go deep on casting in Part Two; for now, remember: the box can change, and you are in charge.

- *Note: Dynamic typing: same box, new type of value*

````python
x = 4
x = "Sally"
print(x)
````

````python
x = str(3)
y = int(3)
z = float(3)
print(x)
print(y)
print(z)
````

## Variable Names: The Rules

Every variable needs a name, and Python has strict rules about names. Rule one: a name must start with a letter, or with the underscore character. Never with a number. Rule two: after that first character, a name may contain only letters, numbers, and underscores. No spaces, no dashes, no other symbols. Rule three: names are case sensitive, and we will prove that in the next segment. Rule four: a name cannot be one of Python's reserved keywords, the words the language itself uses, like if, for, and class. Those four rules are the whole law: everything inside them is allowed, and everything outside them is an error. Follow those rules, and all of these names are legal. myvar. my underscore var, with an underscore in the middle. Underscore my underscore var, starting with an underscore. myVar, with a capital V in the middle. MYVAR, all capitals. And myvar2, letters followed by a number at the end. Six legal names, and here every one of them holds the same value: John. Print them all in one line, and you get John, six times over. Legal is one thing, though; readable is another. A good name describes what the box holds: age, price, user underscore name. A variable called x tells the reader nothing; a variable called total underscore price tells the whole story. Your future self will thank you.

- **RULE 1** — Start with a letter or _ — never a number
- **RULE 2** — Only letters, numbers, _ — no spaces or dashes
- **RULE 3** — Case-SENSITIVE: a and A are different
- **RULE 4** — Never a keyword: if, for, class...

````python
myvar = "John"
my_var = "John"
_my_var = "John"
myVar = "John"
MYVAR = "John"
myvar2 = "John"
print(myvar, my_var, _my_var, myVar, MYVAR, myvar2)
````

## Case Matters + Illegal Names

First, the promised proof that names are case sensitive. Lowercase a equals four. Capital A equals Sally. They look like twins, but to Python they are two completely different variables. Print lowercase a, and you get four. Print capital A, and you get Sally. The number four never touched the capital box. One letter's difference in the name means a different box in memory, and mixing them up is a classic beginner bug. Now the illegal names, so you recognise their errors when you meet them. Two myvar, starting with a number: SyntaxError, invalid decimal literal. My dash var: Python reads the dash as a minus sign, and the assignment collapses into a SyntaxError too. And my space var, with a space in the middle: SyntaxError, invalid syntax, because a name must be one unbroken word. Do not memorise these errors; just know that when Python points at a name, one of the four rules was probably broken. Finally, naming style for longer names. Camel case: each new word starts with a capital, like myVariableName. Pascal case: even the first word is capitalised, like MyVariableName. Snake case: words separated by underscores, like my underscore variable underscore name. Snake case is the Python community favourite: clear, calm, and easy to read. Pick snake case for your variables in this course, and your code will look like real Python from day one.

````python
a = 4
A = "Sally"
print(a)
print(A)
````

## Assign Many Values at Once

Python lets you assign values to several variables in a single line, and you will use this trick daily. Watch this line: x, y, z equals Orange, Banana, Cherry. Python matches them in order. The first variable gets the first value: x becomes Orange. The second gets the second: y becomes Banana. And the third gets the third: z becomes Cherry. Print them, and there they are: Orange, Banana, Cherry. Order is everything here: swap two values, and the boxes swap too. One important check: the number of variables must match the number of values, three and three here, or Python raises an error. It is a small line that replaces three separate assignments, and it keeps related values together where they belong. And here is the mirror trick: one value, many variables. x equals y equals z equals Orange, all in one line. Now all three boxes hold Orange. Print any of them: Orange, Orange, Orange. This chained form is perfect when several variables should all start from the same starting value.

- *Note: Variables count MUST match values count*

````python
x, y, z = "Orange", "Banana", "Cherry"
print(x)
print(y)
print(z)
````

````python
x = y = z = "Orange"
print(x)
print(y)
print(z)
````

## Unpack a Collection

A close cousin of multiple assignment is unpacking. Sometimes your values arrive already packed inside a collection. Here is a list called fruits, holding apple, banana, and cherry. A list is Python's built-in way to store several values in order, and Part Three of this course is devoted to it. You can unpack it straight into variables, in one line: x, y, z equals fruits. Python opens the list and deals the values out in order. x becomes apple. y becomes banana. z becomes cherry. Print them, and there they are. Same matching rule as before: the list must hold exactly as many values as you have variables, no more, no fewer. Unpacking is how Python programmers pull data out of collections elegantly. You will see it everywhere, especially when a function hands you several results at once. Remember the shape of it: variables on the left, collection on the right.

````python
fruits = ["apple", "banana", "cherry"]
x, y, z = fruits
print(x)
print(y)
print(z)
````

## Output Variables: Comma vs Plus

You already know print can output a single variable. If x holds the sentence Python is awesome, then print x shows the whole sentence, exactly as stored. Print can also output several variables at once, and there are two ways to do it. Way one: separate them with commas. x, y, and z hold Python, is, and awesome. Print all three with commas, and the result reads beautifully: Python is awesome. Print puts a space between comma separated values for you. Way two: join the strings yourself with plus signs. Print x plus y plus z. Now Python glues the text together with nothing in between. The result is one squashed word: Pythonisawesome. Same variables, two very different results. The comma is the friendly default: it adds the spaces. The plus is a manual join: powerful, but what you join is exactly what you get. Most of the time, especially while learning, reach for the comma first.

- *Note: comma → spaces added for you*
- *Note: plus → glued exactly as-is*

````python
x = "Python"
y = "is"
z = "awesome"
print(x, y, z)
````

````python
print(x + y + z)
````

## The Plus Trap: Number + Text

One warning about the plus way, because every beginner meets this error. The plus sign means different things for different types. For strings, plus joins text together. For numbers, plus adds values. So what happens when you try to add a number and a string? x is five, a number, and y is John, a string. Print x plus y. Crash: TypeError, unsupported operand type for plus: int and str. Python refuses to guess whether you meant mathematics or text. Honestly, that strictness protects you: silently guessing wrong would produce far stranger bugs than a clear error. Fair enough, but look: the comma way handles the same job without blinking. Print x, comma, y, and you get five John, side by side. The comma does not care about types at all: it simply prints each value, with a space between them. So the rule of thumb is simple. Outputting a mixed bag of values? Use commas. Joining pure strings on purpose? Plus is your tool. That one habit will save you from the most common beginner error in Python.

- *Note: Mixed types → use COMMAS*
- *Note: Pure strings → PLUS joins them*

````python
x = 5
y = "John"
print(x + y)
````

````python
print(x, y)
````

## Global and Local Variables

Variables have a home territory, called their scope. A variable created outside any function, at the top level of your file, is a global variable. It belongs to the whole program, and it can be used everywhere: outside functions and inside them. Look: x equals awesome, created outside. Inside the function, print reads that same global x. The program prints: Python is awesome. But now the twist. Create a variable with the same name inside the function: inside, x equals fantastic. This inner x is a local variable: born inside the function, and alive only inside the function. Inside the function, print gives: Python is fantastic. Then the function ends, and the local x vanishes. Outside, print reads the untouched global variable: Python is awesome. Same name, two boxes, two territories. The local one shadows the global one while the function runs, and never harms it. This design is a gift: a function can use short, convenient names inside, without ever disturbing the rest of your program.

- *Note: GLOBAL = created outside, usable everywhere*
- *Note: LOCAL = created inside, lives only there*

````python
x = "awesome"

def myfunc():
  print("Python is " + x)

myfunc()
````

````python
x = "awesome"
def myfunc():
  x = "fantastic"
  print("Python is " + x)
myfunc()
print("Python is " + x)
````

## The global Keyword

Sometimes, though, a function genuinely needs to create or change a global variable. Python's answer is the global keyword. Write the word global, then the variable name, at the top of the function. You are telling Python: this name means the global box, not a new local one. First example: the function creates x using the global keyword, and sets it to fantastic. No global x existed before the function ran. But when the function finishes, x exists everywhere, and print outside gives: Python is fantastic. Second example: a global x already holds the value awesome. The function declares global x, and reassigns it to fantastic. This time the real global box itself is changed. Print afterwards gives: Python is fantastic. Use the global keyword sparingly. Programs are easier to follow when functions mostly mind their own local business. A program where every function reaches out and rewrites global boxes becomes very hard to predict. But when you need it, it is decisive.

- *Note: global x = 'use the outside box, not a new one'*

````python
def myfunc():
  global x
  x = "fantastic"

myfunc()
print("Python is " + x)
````

````python
x = "awesome"

def myfunc():
  global x
  x = "fantastic"

myfunc()
print("Python is " + x)
````
