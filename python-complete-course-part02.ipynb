## Data Types: The Families

Every value in Python belongs to a family, and that family is called its data type. The family you know best already is text: str, short for string, anything written inside quotes. Numbers come in three kinds: int for whole numbers, float for decimals, and complex for numbers with an imaginary part. Sequence types hold ordered collections: list, tuple, and range. The mapping type is the dictionary, dict for short, which stores pairs of keys and values. Set types, set and frozenset, store unique items with no duplicates allowed. The boolean type, bool, has exactly two possible values: True and False. There are also binary types, bytes, bytearray and memoryview, which handle raw data as bytes; you will rarely touch them as a beginner. And finally there is the None type: the single value None, which means, no value at all. Today we go deep on the everyday families: numbers, text, and booleans. The collections each get their own parts later in this series, starting with lists and tuples in Part Three. Why care about families at all? Because the type decides the rules: what a value can do, and what Python will refuse to do with it.

## Ask Python: type()

You do not have to guess a value's family: you can ask Python directly. The type function is our inspector: feed it any value, and it names the family. x equals five, and type answers: class int. A greeting written in quotes: class str. A price with a decimal point, twenty point five: class float. And one j, a number written with the letter j: class complex, the imaginary family. Lock in the pattern: quotes mean text, a decimal point means float, a plain whole number is int, and a trailing j means complex. Programmers read these answers constantly, because a value's type decides what you are allowed to do with it. Keep this inspector close: we will use it all video to check what our boxes really hold.

```python
x = 5
print(type(x))
greeting = "Hello World"
print(type(greeting))
price = 20.5
print(type(price))
z = 1j
print(type(z))
```

## Setting the Type on Purpose

Normally the value decides its own type: write five, and Python makes an int. But you can also order a type in advance, by calling the type's own constructor function. str of Hello World builds a string on purpose. int of twenty builds a whole number. float of twenty point five builds a decimal. Ask type about all three, and it confirms: str, int, float. The collection constructors are just as direct. list, wrapped around a tuple of three fruits, builds the list: apple, banana, cherry. And dict, with name equals John and age equals thirty-six, builds the pair box: name John, age thirty-six. Pocket these constructor names: str, int, float, list, dict. Very soon they become our casting tools, converting values from one type to another.

```python
x = str("Hello World")
y = int(20)
z = float(20.5)
print(type(x))
print(type(y))
print(type(z))
```

```python
fruits = list(("apple", "banana", "cherry"))
print(fruits)
```

```python
person = dict(name="John", age=36)
print(person)
```

## Data Types: The Full Family

Before we leave data types, meet the rest of the family in one photo call. The sequence family first: square brackets build a list. Round brackets build a tuple: same fruits, different family. And range of six is Python's counting machine: it produces numbers on demand, and we will drive it in Part Five. Curly brackets with pairs inside build a dict. Curly brackets with single items build a set. And frozenset is the set's frozen twin: same uniqueness, but it can never change. True by itself is a bool, the truth family from later today. Text written with a b in front becomes bytes: raw data, not readable text. And a variable holding nothing at all, None, reports the NoneType family. That is the complete family photo: every built-in data type on the W3Schools map.

## Numbers: int, float, complex

Now the number family properly: three siblings, int, float, and complex. An int is a whole number, positive or negative, with no decimal point, and with unlimited length. This giant has thirty digits, and Python handles it exactly, digit for digit. A float carries a decimal point: f equals thirty-five e three. That e notation means times ten to the power of three, so f prints as thirty-five thousand point zero. Floats can be huge and negative too: minus eighty-seven point seven e one hundred prints in the same e notation. Complex numbers carry an imaginary part, written with a j: c equals three plus five j, and it prints in brackets as three plus five j. You will rarely need complex numbers in everyday code, but scientists and engineers live on them. One warning to remember forever: floats are approximations. Add point one and point two, and the answer prints with a long tail ending in four, not a clean point three: a tiny rounding ghost that lives in every programming language. One last contrast: integers in many other languages overflow and break at about two billion; Python ints simply grow as large as your memory allows.

```python
big = 123456789012345678901234567890
print(big)
```

```python
f = 35e3
print(f)
neg = -87.7e100
print(neg)
```

```python
c = 3+5j
print(c)
print(0.1 + 0.2)
```

## Convert + A Dice Roll

Numbers can change family. This is type conversion, done with int, float and complex as converter functions. float of one wraps the whole number as a decimal: one point zero. int of two point eight chops the decimal away: two. Int never rounds; it truncates. And complex of one gives one plus zero j: a real part and an empty imaginary part. Strings that look like numbers convert too: int of the text three gives the number three. But complex numbers refuse to become int or float: Python stops with a TypeError. The error explains it plainly: the int function wants a string or a real number, not a complex. Finally, Python can roll dice for you, with the random module. Import random, fix a seed of forty-two so this lesson is repeatable, and ask randrange for a number from one up to, but not including, ten. The roll lands on two: run it without the seed and the dice lands somewhere new every time. And remember the mechanics of conversion itself: it always builds a brand new value, so the original variable keeps its old type until you store the result back.

```python
a = float(1)
print(a)
b = int(2.8)
print(b)
```

```python
c = complex(1)
print(c)
s = int("3")
print(s)
print(int(3+5j))
```

```python
import random
random.seed(42)
print(random.randrange(1, 10))
```

## Casting: Conversion with Intent

Casting is conversion with intent: you name the type you want, and Python rebuilds the value in that type. Start with int casting, three ways. int of one is one; int of two point eight is two, the decimals chopped off; int of the string three is three. Float casting is just as obedient: float of one is one point zero, float of the string three is three point zero, float of the string four point two is four point two. str casting turns anything into text: str of two gives the text two, and str of three point zero gives the text three point zero. Why does casting matter so much? Because data usually arrives as text: typed by a user, read from a file, sent by a website. And text glues instead of adding: the string three plus the string three prints as three three, two characters stuck together. Cast first, though, and int of three plus int of three is six: real addition, real maths. One honest warning before we move on: casting is not magic. int of the word hello cannot work, because hello is not a number in disguise: Python stops with a ValueError, invalid literal for int. Hold onto casting: in Part Nine, everything a user types arrives as a string, and casting is how that text becomes numbers you can calculate with.

```python
print(int(1))
print(int(2.8))
print(int("3"))
```

```python
print(float(1))
print(float("3"))
print(float("4.2"))
```

```python
print(str(2))
print(str(3.0))
print("3" + "3")
print(int("3") + int("3"))
```

```python
print(int("hello"))
```

## Strings: Text in Quotes

From numbers to the other everyday family: strings, Python's text. A string is text in quotes, and single quotes and double quotes mean exactly the same thing. Hello in double quotes, Hello in single quotes: identical output, so choose one style and stay consistent. Need text across several lines? Use triple quotes: three quotes to open, three to close, and the line breaks inside are kept. Print the poem, and its three lines appear exactly as written. Now the idea that unlocks all text work: a string is an array of characters. Square brackets pick one character by position, and counting starts at zero: position one of Hello, World is the letter e. Position zero would be the capital H: burn that into memory, counting starts at zero everywhere in Python. The len function counts characters: Hello, World, comma and space included, is thirteen long. And the word in checks whether a piece lives inside the text: is World in a? True. Keep that little word in mind; it returns as an operator later today.

```python
print("Hello")
print('Hello')
```

```python
poem = """Roses are red,
Violets are blue,
Sugar is sweet"""
print(poem)
```

```python
a = "Hello, World!"
print(a[1])
print(len(a))
print("World" in a)
```

## Slicing Strings

Slicing cuts a piece out of a string: give a start position and a stop position, separated by a colon. Our patient for the whole string section: b equals Hello, World. b, from two to five. Positions two, three and four are taken: the letters l, l, o. The stop position itself is never included: five is where we stop, not a character we take. Out comes llo. Leave the start empty and Python begins at the very beginning: colon five gives Hello. Leave the stop empty and it runs to the very end: from two onwards, llo, World! Negative positions count from the end of the string: minus one is the very last character. So from minus five to minus two carves out orl, the middle of the word World, counted backwards. And notice: b itself never changes. Slicing hands you a new string; the original is untouchable, and that idea returns in the next segment. File this syntax away carefully: in Part Three you will slice lists and tuples with exactly the same square brackets and colons.

- *Note: [start : stop] — the stop is NOT included*

```python
b = "Hello, World!"
print(b[2:5])
print(b[:5])
print(b[2:])
print(b[-5:-2])
```

## Modify Strings

Strings cannot be changed in place, but they can be transformed: every modify method hands you a brand new string. upper shouts the whole text into capitals: HELLO, WORLD. lower whispers it back down: hello, world, in small letters. Real-world text arrives messy, so meet strip: it removes whitespace from both ends. Our raw string has a space at each end; strip cleans it to a tidy Hello, World. replace swaps every occurrence of one piece for another: replace H with J, and Hello becomes Jello. Same method, same rule: the original txt is untouched; you receive a new string. split does the opposite of joining: it cuts a string into a list of pieces at a separator. Split Hello, World at the comma, and you get a list of two pieces: Hello, and space World. That second piece keeps its leading space: split cuts exactly where you tell it, nothing more. upper, lower, strip, replace, split: five transforms you will use every single day.

```python
txt = "Hello, World!"
print(txt.upper())
print(txt.lower())
```

```python
raw = " Hello, World! "
print(raw.strip())
print(txt.replace("H", "J"))
```

```python
print(txt.split(","))
```

## Join + Format Strings

Two ways to build text remain: joining strings, and formatting values into them. Concatenation joins with plus: a is Hello, b is World, and a plus b glues them with nothing in between: HelloWorld. Want the space? Add it yourself: a, plus a quoted space, plus b, gives Hello World. Now the professional tool: the f-string. Write the letter f before the quotes, and curly braces inside become slots that Python fills with values. age is thirty-six, and the f-string drops it straight into the sentence: My name is John, I am thirty-six. A placeholder can also carry formatting orders: price is fifty-nine, and colon point two f orders exactly two decimal places. The sentence prints: The price is fifty-nine point zero zero dollars. And a placeholder can even hold a calculation: twenty times three is worked out inside the braces and lands as Total: sixty.

```python
a = "Hello"
b = "World"
print(a + b)
print(a + " " + b)
```

```python
age = 36
txt = f"My name is John, I am {age}"
print(txt)
```

```python
price = 59
txt = f"The price is {price:.2f} dollars"
print(txt)
print(f"Total: {20 * 3}")
```

## Escape Characters

A small problem with big annoyance: how do you put a quote inside a string that is itself written with quotes? If Python met a bare quote in the middle, it would end the string early and choke on everything after it. The escape character rescues you: a backslash before a quote tells Python, this quote is content, not the end. So the famous Vikings sentence prints with its inner quotes perfectly intact. The backslash leads a small family of tricks: backslash n inside a string starts a brand new line. One print of Hello backslash n World lands on two lines of output. And to print a backslash itself, double it: two backslashes in code become one on screen, which is how Windows paths like C backslash Users backslash Fahim are written. Two more members to recognise on sight: backslash t leaves a wide tab gap, and backslash quote rescues a quote inside a string written with that same quote. One rule covers the whole family: a backslash changes the meaning of the character that follows it.

```python
txt = "We are the so-called \"Vikings\" from the north."
print(txt)
```

```python
print("Hello\nWorld!")
print("C:\\Users\\Fahim")
```

## String Methods: The Toolbox

Strings arrive carrying a toolbox: built-in methods, ready-made mini-programs that inspect or transform text. A method is called with a dot: the string, a dot, the method name, and brackets. capitalize makes the first character upper case and settles the rest into lower case: Hello, welcome to python. count reports how many times a piece appears: the letter o shows up four times in our sentence. find hunts for a piece and reports where it starts: welcome begins at position seven. And isdigit answers a yes-or-no question about our code string, two zero two four: is every character a digit? True. Here is the honest limit of one segment: Python ships with more than forty built-in string methods. Names worth recognising when you meet them in real code: title, which capitalises every word, plus join, startswith, endswith, and swapcase. ## Booleans: True or False

Booleans are the values of truth itself: either True or False, both written with a capital first letter. Every comparison you write produces one: is ten greater than nine? True. Is ten equal to nine? False. Is ten less than nine? False. Booleans are the steering wheel of programming: every if statement you will ever write turns on one. The bool function evaluates any value you hand it: bool of the text Hello is True. bool of the number fifteen is also True. Almost everything is True: any text with something in it, any number except zero. The False family is small enough to memorise: empty text, the number zero, empty collections, None, and False itself. Watch a boolean steer a real program: a holds two hundred, b holds thirty-three, and the if asks, is b greater than a? It is not, so Python takes the else road and prints: b is NOT greater than a. You can also park a boolean in a variable like any other value, and in Part Five these two little words take charge of if statements and loops.

```python
print(10 > 9)
print(10 == 9)
print(10 < 9)
```

```python
print(bool("Hello"))
print(bool(15))
print(bool(""))
print(bool(0))
```

```python
a = 200
b = 33
if b > a:
  print("b is greater than a")
else:
  print("b is NOT greater than a")
```

## Operators 1: Arithmetic

Operators are the verbs of Python: symbols that perform operations on values and variables. First family: arithmetic, the maths operators. Our test values: x equals ten, y equals three. Plus adds: thirteen. Minus subtracts: seven. The star multiplies: thirty. Slash divides, and division always answers as a float: three point three three three three, and the threes go on. Percent is modulus, the remainder after division: ten divided by three leaves a remainder of one. Double star is power: ten to the power of three is one thousand. Double slash is floor division: divide, then chop down to a whole number: three. Seven operators in the family: add, subtract, multiply, divide, remainder, power, and floor divide. Choosing between them is practical: use slash when the exact value matters, like splitting money; use floor division when you need whole units, like full hours hidden inside a pile of minutes. And modulus is the secret weapon of patterns: even or odd numbers, every third item, clocks and calendars all run on remainders.

```python
x = 10
y = 3
print(x + y)
print(x - y)
```

```python
print(x * y)
print(x / y)
```

```python
print(x % y)
print(x ** y)
print(x // y)
```

## Operators 2: Assignment

Assignment operators put values into boxes, and the smart ones update a box using its own current value. x starts at five: print it, five. x plus-equals three is shorthand for x equals x plus three: the box now holds eight. x star-equals two doubles whatever is inside: sixteen. x minus-equals four brings it down to twelve. Every arithmetic operator has a partner like these: slash-equals, percent-equals, floor-divide-equals, and power-equals. They exist because updating a running total or a counter is something programs do a thousand times a day. Read them as a single instruction from now on: take what is in the box, operate, store it back. The family also has bitwise partners, ampersand-equals, pipe-equals and the shift-equals pair; you will meet them in advanced code, and the operator reference holds their full list.

```python
x = 5
print(x)
x += 3
print(x)
x *= 2
print(x)
x -= 4
print(x)
```

## Operators 3: Comparison

Comparison operators weigh two values against each other and answer with a boolean. Our scales today: x equals five, y equals three. Run the whole family. Double-equals asks, are they the same? False. Not-equals asks, are they different? True. Greater than: True. Less than: False. Greater-or-equal: True. Less-or-equal: False. Now the beginner trap, so you never fall in: one equals sign assigns, two equals signs compare. x equals five puts five into the box; x double-equals five asks a question about the box. Every answer in that table came from the same two numbers; change y to five and the table flips: equal becomes True, not-equal becomes False, and both greater and less fall silent.

- *Note: = assigns · == compares*

```python
x = 5
y = 3
print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

## Operators 4: Logical

Logical operators combine booleans: and, or, and not, the glue inside every real decision. Take x equals five, and test two claims about it at the same time. and demands that both sides be True: five is greater than three, and five is less than ten: True. or needs only one side to be True: greater than three, or less than four: the first claim alone carries it: True. not flips whatever it touches: not of that true combination gives False. and, or, not: add these to comparisons, and you can express any rule a program will ever need to follow. Real rules sound exactly like this: the user is logged in AND the cart is not empty, OR the user is an admin: three claims, three little operators, one decision.

```python
x = 5
print(x > 3 and x < 10)
print(x > 3 or x < 4)
print(not(x > 3 and x < 10))
```

## Operators 5: Identity + Membership

Two operator families remain, and they answer questions about identity and belonging. x and y are two lists that look identical: apple and banana, written out twice. And z is special: z equals x does not copy the list; it gives the same list a second name. The is operator asks, are these literally the same object? x is z: True. x is y: False. The two lists are twins, but they are separate objects in memory. Yet x double-equals y is True, because their contents match exactly. That is the whole lesson in one line: is compares identity, double-equals compares value. Membership asks about belonging: is banana in x? True. Is pineapple not in x? True again: it is nowhere in the list. You met in earlier, checking text inside a string; it works the same on every collection in Python. And each has an opposite twin for free: is not asks the reverse of is, and not in asks the reverse of in.

```python
x = ["apple", "banana"]
y = ["apple", "banana"]
z = x
print(x is z)
print(x is y)
print(x == y)
print("banana" in x)
print("pineapple" not in x)
```

## Operator Precedence

Last topic of the day: when several operators share one line, who goes first? That pecking order is called precedence. One hundred plus five times three: what does Python answer? Multiplication outranks addition, so Python multiplies first, then adds: one hundred and fifteen. Parentheses beat everything: wrap the addition in brackets and it happens first: one hundred and five times three is three hundred and fifteen. Operators of equal rank settle it left to right: five plus four, minus seven, plus three lands on five. The order to remember: parentheses first, then powers, then multiplication and division, then addition and subtraction. And when you are ever unsure, add parentheses yourself: correct and clear beats clever, every time. The complete precedence table lives in the Python Reference, running from parentheses at the very top down to assignment at the bottom; the four levels you learned today cover almost all real code you will ever read.

```python
print(100 + 5 * 3)
```

```python
print((100 + 5) * 3)
```

```python
print(5 + 4 - 7 + 3)
```
